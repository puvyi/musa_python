```python
def day_of_week():
    user_input=input("Please enter a number between 1 and 7: ")
    n=int(user_input)
    if n==1:
        print("Monday")
    elif n==2:
        print("Tuesday")
    elif n==3:
        print("Wednesday")
    elif n==4:
        print("Thursday")
    elif n==5:
        print("Friday")
    elif n==6:
        print("Saturday")
    else:
        print("Sunday")
```

```python
day_of_week()
```
