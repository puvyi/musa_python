```python
file = open('words.txt', 'r')
```

```python
def is_ordered_word(word):
    for i in range(0,len(word)):
        for j in range(i+1, len(word)):
            if ord(word[i])>ord(word[j]):
                return False
    else:
        return True
                
```

```python
is_ordered_word("abcdasd")
```

```python
for line in file:
    word = line.rstrip()
    if is_ordered_word(word):
        print(word)
```

```python
def uses_all(text, letters):
    uses = True
    for letter in letters:
        if letter not in text:
            uses=False
    return uses
```

```python
uses_all("ammaasdasdxcv", 'ma')
```

```python
for line in file:
    word = line.rstrip()
    if uses_all(word.lower(), 'aei'):
        print(word)
```

```python
def only_uses_these_letters(text, letters):
    for l in text:
        if l not in letters:
            return False
    else:
        return True
```

```python
for line in file:
    word = line.rstrip()
    if only_uses_these_letters(word.lower(), 'ens'):
        print(word)
```

```python
only_uses_these_letters('mydear', 'rydeam')
```

```python
def is_any_forbidden_letter(forbidden_letters, text):
    ans = False
    for letter in text:
        if letter in forbidden_letters:
            ans = True
    return ans
```

```python
is_any_forbidden_letter('zy', 'Vengalappaale')
```

```python
for line in file:
    word = line.rstrip()
    if is_any_forbidden_letter('b', word.lower()):
        print(word)
```

```python
def has_vowel(word):
    vowel = False
    for letter in word.lower():
        if letter in 'aeiou':
            vowel = True
    return vowel
```

```python
for line in file:
    word = line.rstrip()
    if not has_vowel(word):
        print(word)
```
