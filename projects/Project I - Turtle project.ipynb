```python
import turtle
import math
```

```python
t = turtle.Turtle()
```

```python
def square(t, d:int, is_screen_reset:bool= False):
    """
    Draws a square with Turtle
    Parameters: t: turtle, d: side length, is_screen_reset: bool
    returns: None, but draws on the Turtle screen
    """
    if is_screen_reset:
        turtle.resetscreen()
    for i in range(4):
        t.fd(d)
        t.lt(90)
```

```python
square(t, 40)
```

```python
turtle.resetscreen()
for i in range(50, 100, 10):
    square(t, i, True)
    t.lt(360/30)
```

```python
def polygon(t, d, n, is_screen_reset:bool=False):
    """
    Draws a polygon of n sides with Turtle
    Parameters: t: turtle, d: side length, n: number of sides for the polygon, is_screen_reset: bool
    returns: None, but draws on the Turtle screen
    """
    if is_screen_reset:
        turtle.resetscreen()
    for i in range(n):
        t.fd(d)
        t.lt(360/n)
    
    
```

```python
def circle(t, radius, is_screen_reset=False):
    steps = 360/10
    polygon(t, int(math.pi*radius/steps), 36)
```

```python
turtle.resetscreen()
```

```python
for i in range(4):
    circle(t, i*100)
```

```python
polygon(t, 10, 80, True)

```

```python
turtle.resetscreen()
for i in range(20):
    polygon(t, 10, 80)
    t.lt(360/20)
```

```python
# DRAW a STAR
turtle.resetscreen()
```

```python
t.fillcolor("orange")
t.begin_fill()
```

```python
def star(t, a, arms, is_screen_reset=False):
    t.fillcolor("orange")
    t.begin_fill()
    t.lt(60)
    
    for i in range(1, arms+1):
        t.fd(a)
        t.rt(360/arms)
        t.fd(a)
        t.lt(2*360/arms)
    t.end_fill()
        
    
```

```python
for i in range(6):
    star(t, 50, 5)
for i in range(6):
    star(t, 60, 5)
```

```python
# Must always be the last command

turtle.mainloop()
```
